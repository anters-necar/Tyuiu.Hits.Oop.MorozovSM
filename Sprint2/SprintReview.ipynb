```polyglot-notebook
public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    List<string> GetTeamMembers();
}

public class Employee {
    public string name { get; set; }
    public string position { get; set; }
    public float salary { get; set; }

    public virtual void PrintInfo(){
        Console.WriteLine($"Имя: {name} Должность: {position} Зарплата: {salary}");
    }
}

public class Manager : Employee {
    public string department { get; set; }

    public override void PrintInfo(){
        Console.WriteLine($"Имя: {name} Должность: {position} Зарплата: {salary} Отдел: {department}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader {
    public string projectName { get; set; }
    public List<string> teamMembers { get; set; } = new List<string>();

    public override void PrintInfo(){
        Console.WriteLine($"Имя: {name} Должность: {position} Зарплата: {salary} Отдел: {department} Реализуемый проект: {projectName}");
    }

    public string GenerateReport(){
        return $"Отчет по проекту: {projectName}. За проект отвечает {name}.";
    }

    public List<string> GetTeamMembers(){
        return teamMembers;
    }
}

Employee employee = new Employee { name = "Александр", position = "Разработчик", salary = 35000 };
Manager manager = new Manager { name = "Владимир", position = "Менеджер", salary = 50000, department = "Хоз. отдел" };
ProjectManager projectManager = new ProjectManager{
    name = "Дмитрий",
    position = "Руководитель склада",
    salary = 120000,
    department = "Хоз. отдел",
    projectName = "Автоматизация процессов склада",
    teamMembers = new List<string> { "Иван", "Мария", "Сергей" }
};

employee.PrintInfo();
manager.PrintInfo();
projectManager.PrintInfo();
Console.WriteLine(projectManager.GenerateReport());
Console.WriteLine("Состав команды: " + string.Join(", ", projectManager.GetTeamMembers()));
```

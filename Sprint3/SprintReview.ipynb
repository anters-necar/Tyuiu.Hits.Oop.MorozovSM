```polyglot-notebook
abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"{Name}, цена: {Price}";
    }

    public override string ToString() => GetInfo();

    public static bool operator ==(Component c1, Component c2)
    {
        if (ReferenceEquals(c1, c2)) return true;
        if (ReferenceEquals(c1, null) || ReferenceEquals(c2, null)) return false;
        return c1.Name == c2.Name && c1.Price == c2.Price;
    }

    public static bool operator !=(Component c1, Component c2)
    {
       if (ReferenceEquals(c1, c2)) return false;
        if (ReferenceEquals(c1, null) || ReferenceEquals(c2, null)) return true;
        return c1.Name != c2.Name || c1.Price != c2.Price;
    }

    public override bool Equals(object obje)
    {
        var other = obje as Component;
        return this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}

interface IInstallable
{
    void Install();
}

class CPU : Component, IInstallable
{
    public CPU(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"CPU: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Устанавливаются драйвера для процессора {Name}.");
}

class RAM : Component, IInstallable
{
    public RAM(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"RAM: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Устанавливаются драйвера для оперативной памяти {Name}.");
}

class GPU : Component, IInstallable
{
    public GPU(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"GPU: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Устанавливаются драйвера для видеокарты {Name}.");
}

class HDD : Component, IInstallable
{
    public HDD(string name, decimal price) : base(name, price) { }
    public override string GetInfo() => $"HDD: {base.GetInfo()}";
    public void Install() => Console.WriteLine($"Устанавливаются драйвера для HDD {Name}.");
}

class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удалён из компьютера.");
        }
        else
        {
            Console.WriteLine($"Компонент {component.Name} не найден.");
        }
    }

    public decimal GetTotalPrice()
    {
        return components.Sum(c => c.Price);
    }

    public void ShowComponents()
    {
        Console.WriteLine("Состав компьютера:");
        foreach (var comp in components)
        {
            Console.WriteLine(comp);
        }
    }
}
Computer computer = new Computer();

CPU cpu = new CPU("Ryzen 7 2700", 10000);
RAM ram = new RAM("Kingston 16GB", 8000);
GPU gpu = new GPU("NVIDIA RTX 1070ti", 20000);
HDD hdd = new HDD("Toshiba 2TB", 5000);
computer.AddComponent(cpu);
computer.AddComponent(ram);
computer.AddComponent(gpu);
computer.AddComponent(hdd);

Console.WriteLine("Что в компьютере:");
computer.ShowComponents();
Console.WriteLine($"Общая стоимость: {computer.GetTotalPrice()}");

Console.WriteLine("удаление компонента");
computer.RemoveComponent(ram);

Console.WriteLine("Что в компьютере после удаления:");
computer.ShowComponents();
Console.WriteLine($"Общая стоимость после удаления: {computer.GetTotalPrice()}");

Console.WriteLine("Сравнение компонентов");
CPU cpu2 = new CPU("Ryzen 7 2700", 10000);
CPU cpu3 = new CPU("Ryzen 5 2600", 20000);
Console.WriteLine($"CPU1 == CPU2: {cpu == cpu2}");
Console.WriteLine($"CPU1 == CPU3: {cpu == cpu3}");
```
